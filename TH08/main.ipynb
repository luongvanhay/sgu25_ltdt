Mã giả:

void greedyColoring(vector<vector<int>>&adj)

        gán n=size của adj
        tạo vector<int>result(n,-1)
        khởi tạo phần tử đầu tiên mảng result=0
        tạo mảng bool available(n,false)

        lặp u =1, u<n
        với mọi phần tử trong adj[u]
        nếu result[phần tử trong adj]!=-1
        gán available[result[phần tử trong adj]]=true;

        tạo int cr
        với cr=0;cr<n
        nếu available cr ra sai thì break;

        gán result[u]=cr;

        lặp mọi phần tử trong adj[u]
        nếu result[neighbour:adj[u]]
        available[result[neighbour]]=false;


    tìm màu lớn nhất trong mảng result

    tạo mảng nhóm màu 
    với u=0, u<n
    nhóm màu[result[u]].push_back(u+1);
